## Task 1

```python
score = int(input("Enter your exam score: "))

if 100 >= score >= 91:
    grade = "A"
elif score >= 81:
    grade = "B"
elif score >= 71:
    grade = "C"
elif score >= 61:
    grade = "D"
elif score >= 51:
    grade = "E"
elif score >= 0:
    grade = "F"
else:
    grade = "1"

if grade == "1":
    print("Invalid output!")
else:
    print("Your grade is", grade)
```

## Task 2

```python
n = int(input("Enter a positive number: "))

sum_of_evens = 0
num_of_nums_divisible_3 = 0

if n <= 0:
    print("Invalid input!")
else:
    for i in range(1, n + 1):
        if i % 3 == 0 and i % 5 == 0:
            print(" 3 5")
            num_of_nums_divisible_3 += 1
        elif i % 3 == 0:
            print(" 3")
            num_of_nums_divisible_3 += 1
        elif i % 5 == 0:
            print(" 5")
        else:
            print(i)

        if i % 2 == 0:
            sum_of_evens += i

    print("The sum of all even numbers from 1 to n:", sum_of_evens)
    print("How many numbers is divisible by 3:", num_of_nums_divisible_3)
```
